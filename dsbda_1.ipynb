```python
import pandas as pd
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
lrmodel=LinearRegression()
```

```python
x=np.array([4,2,3,5,1,3]).reshape(-1, 1)
y=np.array([3,4,3,5,1,3])
```

```python
lrmodel.fit(x,y)
```

```python
slope=lrmodel.coef_[0]
intercept=lrmodel.intercept_
print("slope :",slope)
print("intercept :",intercept)
print(f"y={intercept}+{slope}x")
```

```python
ypredict=lrmodel.predict(x)
ypredict
```

```python
y
```

```python
nextpr=lrmodel.predict([[6]])
nextpr
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y,color='blue',label='actual')
plt.xlabel('bahubali-1 rating')
plt.ylabel('bahubali-2 rating')
plt.title('actual vs predicted rating')
plt.scatter(x,ypredict,color='red',label='predicted')
plt.plot(x,ypredict,color='green',label='regression line')
plt.legend()
```

```python
from sklearn.metrics import mean_squared_error,accuracy_score,r2_score
mse=mean_squared_error(y,ypredict)
# acc=accuracy_score(y,ypredict)
r2=r2_score(y,ypredict)
# print("accuracy of our model is :",acc)
print("r2_score is :",r2)
print("mean squared error",mse)
```
