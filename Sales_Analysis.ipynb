```python
import pandas as pd
df = pd.read_excel(r"C:\Users\abhis\Downloads\Sales-Data-Analysis.xlsx",header=0)
df.head()
```

```python
df.columns = df.iloc[0]
df.head()
```

```python
df.info()
```

```python
#df=df.drop(columns=["unnamed:0"])
#df = df.drop(columns=["NaN"])
df = df.dropna(axis=1, how="all")
```

```python
df.head()
```

```python
df = df[df["Order ID"] != "Order ID"].reset_index(drop=True)
```

```python
df.head()
```

```python
# #df=df.drop(columns=["Order ID:1"])
# #df = df.drop(columns=["Order ID"])
# df=df.drop(index=1)
# df.head()
```

```python
df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Price"] = df["Price"].astype(float)
df["Quantity"] = df["Quantity"].astype(float)
df["Order ID"] = df["Order ID"].astype(int)
```

```python
df.head()
```

```python
print("----------------------------------------info-----------------------------------------------")
df.info()
print("----------------------------------------isnasum-----------------------------------------------")
df.isna().sum()
print("----------------------------------------describe-----------------------------------------------")
df.describe()
```

```python
df["Revenue"] = df["Price"] * df["Quantity"]
df.head()
```

```python
total_revenue = df["Revenue"].sum()
total_orders = df["Order ID"].nunique()
avg_order_value = total_revenue / total_orders
print(f"total revenue :",total_revenue,"\ntotal orders:", total_orders,"\navg value per order:", avg_order_value)
```

```python
product_revenue = (
df.groupby("Product")["Revenue"]
.sum()
.sort_values(ascending=False)
)
product_revenue
```

```python
product_qty_rev = df.groupby("Product")[["Quantity", "Revenue"]].sum()
product_qty_rev
```

```python
# product_quantity_rev=df.groupby("Product")[["Quantity","Revenue"]].sum()
```

```python
purchase_type_rev = df.groupby("Purchase Type")["Revenue"].sum()
purchase_type_rev
```

```python
purchase_type_rev = df.groupby("Purchase Type")["Revenue"].sum().sort_values(ascending=False)
purchase_type_rev
```

```python
product_channel = pd.pivot_table(
df,
values="Revenue",
index="Product",
columns="Purchase Type",
aggfunc="sum"
)
product_channel
```

```python
manager_revenue = (
df.groupby("Manager")["Revenue"]
.sum()
.sort_values(ascending=False)
)
manager_revenue
```

```python
city_revenue = (
df.groupby("City")["Revenue"]
.sum()
.sort_values(ascending=False)
)
city_revenue
```

```python
df["Month"] = df["Date"].dt.to_period("M")
monthly_revenue = df.groupby("Month")["Revenue"].sum()
monthly_revenue
```

```python
high_qty_orders = df[df["Quantity"] > df["Quantity"].quantile(0.99)]
high_qty_orders
```

```python
product_revenue.to_csv("product_revenue.csv")
product_channel.to_csv("product_channel.csv")
manager_revenue.to_csv("manager_revenue.csv")
city_revenue.to_csv("city_revenue.csv")
monthly_revenue.to_csv("monthly_revenue.csv")
```
